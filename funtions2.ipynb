```python
#function for Even/odd
def check_even_odd(number):
    if number % 2 ==0:
        return "Even"
    else:
        return "Odd"
    
    result = check_even_odd(25)
    print(result)
```

```python
#function + loop
def multipilication_table(number):

    for i in range(1, 11):
        print(number, "x", i, "=", number * i)

number = int(input("enter number:"))

multipilication_table(number)
```

```python
#Largest of three numbers
def largest(a, b, c):
    if a >= b and a >= c:
        return a
    elif b >= a and b >=c:
        return b 
    else:
        return c

a = int(input("Enter a:"))
b = int(input("Enter b:"))
c = int(input("Enter c:"))

print("Largest =", largest(a, b, c))
```

```python
#grade calculation
def find_grade(marks):
    if marks >=90:
        return "A"
    elif marks >=80:
        return "B"
    elif marks >=70:
        return "C"
    elif marks >=60:
        return "D"
    elif marks >=40:
        return "E"
    else:
        return "Fail"

marks = int(input("Enter marks:"))

print("Grade =", find_grade(marks))
    
```
